### RecursiveCharacterTextSplitter


The RecursiveCharacterTextSplitter is a document-splitting utility from LangChain.

It is used to break large text into smaller chunks recursively so that they can be processed by LLMs or stored in vector databases for RAG (Retrieval Augmented Generation) systems.


### Why RecursiveCharacterTextSplitter is used

* LLMs have token limits, so large documents must be split.

This splitter:

    * Breaks text into chunks of fixed size
    
    *  Keeps semantic boundaries when possible
    
    * Uses multiple separators recursively
    
* Typical separators: ["\n\n", "\n", " ", ""]

### It tries splitting by:

    1 Paragraph
    2 Line
    3 Word
    4 Character

So the text structure is preserved as much as possible.

| Parameter         | Meaning                                    |
| ----------------- | ------------------------------------------ |
| `chunk_size`      | Maximum size of each chunk                 |
| `chunk_overlap`   | Overlap between chunks                     |
| `separators`      | Rules used to split text                   |
| `length_function` | Function to measure length (default = len) |


```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)

final_documents=text_splitter.split_documents(docs)

len(final_documents)
```

### CharacterTextSplitter


* CharacterTextSplitter splits text into fixed-size chunks purely based on character count —

* it does not try to preserve sentence or paragraph boundaries like RecursiveCharacterTextSplitter does.

* What CharacterTextSplitter Does

      * Splits text based on a specific separator

      * Creates fixed-size chunks

      * Supports overlapping chunks

| Parameter         | Description                   |
| ----------------- | ----------------------------- |
| `separator`       | Character used to split text  |
| `chunk_size`      | Maximum characters in a chunk |
| `chunk_overlap`   | Overlap between chunks        |
| `length_function` | Function to measure length    |


```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter=CharacterTextSplitter(separator="\n\n",chunk_size=100,chunk_overlap=20)

text_splitter.split_documents(docs)
```

```python

speech=""

with open("speech.txt") as f:
    speech=f.read()

text_splitter=CharacterTextSplitter(chunk_size=50,chunk_overlap=20)

text=text_splitter.create_documents([speech])

print(text[0])
print(text[1])
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="",     
    chunk_size=100,
    chunk_overlap=20
)
docs = text_splitter.create_documents([speech])

for i, d in enumerate(docs[:5]):
    print(f"Chunk {i}: {len(d.page_content)} chars")
    print(d.page_content)
```

| Feature               | CharacterTextSplitter | RecursiveCharacterTextSplitter |
| --------------------- | --------------------- | ------------------------------ |
| Splitting logic       | Fixed separator       | Hierarchical, smart            |
| Meaning preservation  | ❌ Low                 | ✅ High                         |
| Chunk quality for RAG | ⚠️ Okay               | 🔥 Excellent                   |
| Speed                 | Fast                  | Slightly slower                |
| Best for              | Clean/structured text | PDFs, articles, docs, RAG      |
