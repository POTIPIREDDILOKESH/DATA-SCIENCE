## Explore generation parameters



### Output length

✅ Output length affects both cost and performance.

⏳ More tokens = more computation, higher latency, more energy use, and higher cost.

⚙️ Use the <b>max_output_tokens</b> parameter in the Gemini API to set a limit on how many tokens the model can generate.

🚫 This parameter only stops generation after the limit — it doesn’t make the output shorter or more concise by itself.

🧠 To get a complete and meaningful answer within the limit, use prompt engineering (clear, specific, focused prompts).


### Temperature

🌡 Temperature controls how random or creative the model’s output is.

🔥 Higher temperature → more randomness and diversity in responses.

🧊 Lower temperature → more focused, predictable, and deterministic output.

⚙️ Temperature = 0 → greedy decoding (model always picks the most probable next token).

🎲 Temperature doesn’t guarantee randomness — it only nudges the output style.

### Top-K and top-P

🎛 Top-K and Top-P also control output diversity like temperature.

🔢 Top-K = number of most probable tokens to choose from.

        Example: topK = 1 → greedy decoding (always picks the top token).

🎯 Top-P = cumulative probability threshold for candidate tokens.

        Example: topP = 0 → greedy decoding; topP = 1 → includes all tokens.
        

```python
Assume a language model predicts the next word after:

“I love eating”

It outputs probabilities like this:
```

| Token    | Probability |
| -------- | ----------- |
| pizza    | 0.40        |
| biryani  | 0.25        |
| pasta    | 0.15        |
| icecream | 0.10        |
| salad    | 0.05        |
| sushi    | 0.03        |
| burger   | 0.02        |



Top-K = 1 : Always selects pizza

This is greedy decoding.


```python
Top-K = 3: 
            [pizza (0.40), biryani (0.25), pasta (0.15)]

            All others are removed.
            
            Now we renormalize probabilities:
            
            New total = 0.40 + 0.25 + 0.15 = 0.80

New probabilities:

            pizza → 0.40 / 0.80 = 0.50
            
            biryani → 0.25 / 0.80 = 0.31
            
            pasta → 0.15 / 0.80 = 0.19
            
            Now the model randomly picks among these 3.

👉 Output becomes more diverse than greedy.
```

Top-P (Nucleus Sampling)

Definition: Choose the smallest set of tokens whose cumulative probability ≥ P.

1. Top-P = 0.50

    Add tokens from highest probability until sum ≥ 0.50:

        pizza = 0.40

        pizza + biryani = 0.65  (stop here)

        Allowed tokens:[pizza, biryani]

2. Top-P = 0.80

        pizza = 0.40
        
        biryani = 0.65
        
        pasta = 0.80  (stop)

        Allowed tokens: [pizza, biryani, pasta]


⚙️ When both are set, the model applies them in order:

    1.Filter top-K tokens.

    2.Apply top-P probability cutoff.

    3.Sample from the remaining tokens using the temperature value.

🧠 Together, they balance creativity vs. control in the generated output.
